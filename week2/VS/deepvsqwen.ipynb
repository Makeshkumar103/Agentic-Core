```python
groq = Groq()
client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=os.getenv("OPENROUTER_API_KEY"),
)

# groq_url = "https://api.groq.com/openai/v1"
# groq = OpenAI(api_key=groq_api_key, base_url=groq_url)

models = client.models.list()
for m in models:
    print(m.id)
```

```python
deepseek = "deepseek/deepseek-v3.2"
qwen3 = "qwen/qwen3.7-flash"

deep_system = "You are a chatbot who is very argumentative; \
you disagree with anything in the conversation and you challenge everything, in a snarky way."

qwen_system = "You are a very polite, courteous chatbot. You try to agree with \
everything the other person says, or find common ground. If the other person is argumentative, \
you try to calm them down and keep chatting."

deep_messages = ["Hi there"]
qwen_messages = ["Hi"]
```

```python
def call_deep():
    messages = [{"role": "system", "content": deep_system}]
    for deep, qwen in zip(deep_messages, qwen_messages):
        messages.append({"role": "assistant", "content": deep})
        messages.append({"role": "user", "content": qwen})
    response = client.chat.completions.create(model=deepseek, messages=messages)
    return response.choices[0].message.content
```

```python
call_deep()
```

```python
def call_qwen():
    messages = [{"role": "system", "content": qwen_system}]
    for deep, claude_message in zip(deep_messages, qwen_messages):
        messages.append({"role": "user", "content": deep_messages})
        messages.append({"role": "assistant", "content": qwen_messages})
    messages.append({"role": "user", "content": deep_messages[-1]})
    response = client.chat.completions.create(model=qwen3, messages=messages)
    return response.choices[0].message.content
```
